# **Electron Density Analysis and Theoretical Implications**

## **1. Introduction**

### **Physics Background**
The study of electron density in the magnetosphere is crucial for understanding space plasmas and their interactions with Earth's magnetic field. The electron plasma frequency (FPE) is a fundamental quantity in plasma physics, given by:

\[
 f_{pe} = 8980 \times \sqrt{n_e}
\]

where ne n_e  is the electron density (in cm⁻³) and fpe f_{pe}  is in Hz. By analyzing the variations in ne n_e , we can infer changes in the plasma environment due to solar activity, geomagnetic storms, and space weather phenomena.

---


## **2. Dataset and Methodology**

### **Data Sources**
The dataset consists of experimental electron density values derived from the IMAGE RPI Passive Plasma Emissions Data (2001-2005). We also generate a theoretical dataset with a 10% variation to model potential fluctuations.

### **Processing Steps:**
1. Convert FPE (kHz) to electron density (cm⁻³) using ne=(FPEHz/8980)2 n_e = (FPE_{Hz} / 8980)^2 .
2. Generate a theoretical dataset by introducing controlled random variations.
3. Conduct statistical analysis (mean, standard deviation, percentiles).
4. Visualize the data using scatter plots, histograms, contour maps, and box plots.

---

```python
import pandas as pd
import numpy as np
import plotly.express as px
import plotly.graph_objects as go

# Load dataset
file_path = "/kaggle/input/image-rpi-passive-plasma-data-emissions/IMAGE RPI electron density from passive plasma emissions.csv"  # Update with the actual file path
df = pd.read_csv(file_path)

# Convert FPE (kHz) to electron density (cm⁻³)
df['FPE_Hz'] = df['FPE'] * 1e3  # Convert kHz to Hz
df['ne'] = (df['FPE_Hz'] / 8980) ** 2

# Replace inf values with NaN
df.replace([np.inf, -np.inf], np.nan, inplace=True)

# Drop rows with NaN values
df.dropna(inplace=True)

# Generate a theoretical dataset based on a model
np.random.seed(42)
df_theoretical = df.copy()
df_theoretical['ne'] = df['ne'] * (1 + 0.1 * np.random.randn(len(df)))  # Add 10% random variation
```

```python
# Statistical properties
stats = df.groupby('Year')['ne'].agg(['mean', 'std', 'median', lambda x: np.percentile(x, 25), lambda x: np.percentile(x, 75)])
stats.columns = ['Mean', 'Std Dev', 'Median', '25th Percentile', '75th Percentile']
print(stats)
```

## **3. Interpretation of Graphs**


### **Scatter Plot: Radial Distance vs. Magnetic Latitude (MLAT) Colored by Electron Density**
- **Meaning:** Shows the distribution of electron density across different radial distances and latitudes in Earth's magnetosphere.
- **Implications:** Helps identify regions of high plasma density, such as the plasmasphere and radiation belts.

```python
# Scatter plot: Radial Distance vs MLAT, colored by ne
fig = px.scatter(df, x='Radial_Distance', y='MLAT', color='ne', color_continuous_scale='viridis', title='Electron Density vs Radial Distance and MLAT')
fig.update_layout(coloraxis_colorbar=dict(title='Electron Density (cm⁻³)'))
fig.show()
```

### **Histogram of Electron Density Distribution**
- **Meaning:** Provides a statistical overview of how electron density is distributed.
- **Implications:** Useful for modeling the likelihood of specific density values occurring in the magnetosphere.

```python
# Histogram of electron density
fig = px.histogram(df, x='ne', nbins=50, title='Distribution of Electron Density', color_discrete_sequence=['blue'])
fig.update_layout(xaxis_title='Electron Density (cm⁻³)', yaxis_title='Frequency')
fig.show()
```

### **Yearly Variation of Electron Density (Box Plot)**
- **Meaning:** Displays variations over different years, revealing trends influenced by solar activity.
- **Implications:** Correlates with the solar cycle, helping predict space weather effects on satellites and communication systems.

```python
# Yearly variation analysis
fig = px.box(df, x='Year', y='ne', title='Yearly Variation of Electron Density', color='Year')
fig.update_layout(yaxis_type='log', xaxis_title='Year', yaxis_title='Electron Density (cm⁻³)')
fig.show()
```
